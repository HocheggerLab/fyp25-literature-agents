# Multi-Model Comparison Study Demo

This notebook demonstrates how to run a comprehensive model comparison study using the FYP25 Literature Agents toolkit.

## Overview

This study will:
1. Search PubMed for gene-cancer abstracts
2. Analyze the same abstracts with **5 different LLM providers**:
   - OpenAI GPT-5 Mini
   - Anthropic Claude 4.5 Haiku
   - Google Gemini 3.0 Pro
   - DeepSeek V3.1-Terminus
   - Groq Llama 4 70B Scout
3. Run each model **5 times** to measure consistency
4. Generate CSV for quantitative comparison

````python
# Import the main function
import json

# For data analysis
import pandas as pd

from fyp25_literature_agents import run_model_comparison
````

## Run the Comparison Study

This will:
- Search PubMed for "PPP2R2A AND cancer"
- Retrieve 20 abstracts
- Analyze with all 5 models
- Run 5 repeats per model
- Total: 20 articles × 5 models × 5 repeats = 500 analyses

**Estimated time:** 10-15 minutes (depends on API response times)

**Estimated cost:** ~$0.39 (see MODEL_COMPARISON_SPEC.md for breakdown)

````python
# Run the comparison study
results = await run_model_comparison(
    search_term="PPP2R2A AND cancer",
    max_results=5,
    num_repeats=2,
    output_dir="model_comparison_results",
    gene="PPP2R2A",  # Optional: auto-extracted if not provided
)
````

## Examine Results

````python
# Print summary
print(f"Search term: {results['search_term']}")
print(f"Gene: {results['gene']}")
print(f"Articles analyzed: {results['num_articles']}")
print(f"Models tested: {len(results['models_tested'])}")
print(f"Repeats per model: {results['num_repeats']}")
print(f"\nOutput directory: {results['output_dir']}")
print(f"CSV file: {results['csv_file']}")
print(f"JSON files: {len(results['json_files'])}")

# Check for failures
if results['failed_analyses']:
    print(f"\n⚠ Failed analyses: {len(results['failed_analyses'])}")
    for failure in results['failed_analyses']:
        print(f"  - {failure['model_name']} repeat {failure['repeat_num']}: {failure['error']}")
else:
    print("\n✓ All analyses completed successfully!")
````

## Load and Explore the CSV Data
